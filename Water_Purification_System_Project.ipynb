```python
turbidity = float(input("Enter turbinity (NTU) : "))
tds = int(input("Enter TDS (ppm) : "))
bacteria = input("Is bacteria present? (True/False): ") == "True"
ph = float(input("Enter pH Level : "))

print("\nInitial water Quality : ")
print(f"Turbidity : {turbidity}")
print(f"TDS : {tds}")
print(f"Bacteria : {bacteria}")
print(f"pH : {ph}")

if turbidity >= 1:
  print("\n→Turbidity > 1 → Action Taken: Sedimentation")
  print("→Turbidity > 1 → Action Taken: Filteration")
  turbidity = 0.5
if tds >= 300 :
  print("→ TDS > 300 → Action Taken: Reverse Osmosis")
  tds =280
if bacteria == True :
  print("Bacteria = True → Action Taken: UV Treatment")
  bacteria = False
if ph < 6.5 :
  print("pH < 6.5 → Action Taken: Add Chemicals")
  ph = 7.2
elif ph > 8.5 :
  print("pH < 8.5 → Action Taken: Add Chemicals")
  ph = 7.2

print("\nWater quality after treatment : ")
print(f"Turbidity : {turbidity}")
print(f"TDS : {tds}")
print(f"Bacteria : {bacteria}")
print(f"pH : {ph}")
print("Water is purified and safe for drinking.")
```
