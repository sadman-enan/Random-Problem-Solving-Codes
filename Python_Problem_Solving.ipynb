```python
number= int(input("Enter number to check :"))
if number % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
month = int(input("Enter month number :"))
match month:
  case 1:
    print("January")
  case 2:
    print("February")
  case 3:
    print("March")
  case 4:
    print("April")
  case 5:
    print("May")
  case 6:
    print("June")
  case 7:
    print("July")
  case 8:
    print("August")
  case 9:
    print("September")
  case 10:
    print("October")
  case 11:
    print("November")
  case 12:
    print("December")
  case _:
    print("Invalid month number")
```

```python
number = int(input("Enter number :"))
if number < 0 :
  print("Invalid Input")
else:
  factorial_number = 1
  for i in range (1, number + 1):
    factorial_number= factorial_number * i

  print(f"Factorial of {number} is : ", factorial_number)
```

```python
number = list(map(int,input("Enter numbers for list : ").split()))
if len(number) == 0:
  print("No inputs found.")
else:
  largest = number[0]
  for i in number:
    if i > largest:
      largest = i

  print("The largest number is : ", largest)
```
